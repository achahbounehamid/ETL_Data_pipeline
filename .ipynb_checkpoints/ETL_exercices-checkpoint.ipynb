### Ex E5 ⭐⭐ — Fonction extract()
Encapsulez tout dans une fonction `extract(q1_path, q2_path, clients_path, produits_path) -> dict`
qui retourne un dictionnaire avec les 4 DataFrames harmonisés et consolidés.
La fonction doit afficher des logs à chaque étape.

```python
def extract(q1_path, q2_path, clients_path, produits_path):
    print("="*60)
    print(" DÉBUT DE L'EXTRACTION")
    print("="*60)
    
    # 1. Chargement des fichiers
    print("\n Étape 1 : Chargement des fichiers CSV")
    commandes_Q1 = pd.read_csv(q1_path)
    print(f" commandes_Q1 chargé : {commandes_Q1.shape[0]} lignes")
    
    commandes_Q2 = pd.read_csv(q2_path)
    print(f" commandes_Q2 chargé : {commandes_Q2.shape[0]} lignes")
    
    clients = pd.read_csv(clients_path)
    print(f" clients chargé : {clients.shape[0]} lignes")
    
    produits = pd.read_csv(produits_path)
    print(f" produits chargé : {produits.shape[0]} lignes")
    
    # 2. Harmonisation des colonnes
    print("\n Étape 2 : Harmonisation des colonnes")
    commandes_Q2.rename(columns={'moyen_paiement': 'mode_paiement'}, inplace=True)
    print(" Colonne 'moyen_paiement' → 'mode_paiement' dans Q2")
    
    commandes_Q1['devise'] = 'EUR'
    print(" Colonne 'devise' ajoutée dans Q1 (valeur par défaut : EUR)")
    
    # 3. Ajout de la colonne source
    print("\n Étape 3 : Ajout de la colonne source")
    commandes_Q1['source'] = 'Q1_2024'
    commandes_Q2['source'] = 'Q2_2024'
    print(" Colonne 'source' ajoutée dans Q1 et Q2")
    
    # 4. Concaténation
    print("\n Étape 4 : Concaténation Q1 + Q2")
    commandes = pd.concat([commandes_Q1, commandes_Q2], ignore_index=True)
    print(f" Concaténation réussie : {commandes.shape[0]} lignes au total")
    
    # 5. Vérification de cohérence
    nb_q1 = commandes_Q1.shape[0]
    nb_q2 = commandes_Q2.shape[0]
    nb_total = commandes.shape[0]
    
    if nb_total == nb_q1 + nb_q2:
        print(f" Cohérence vérifiée : {nb_q1} + {nb_q2} = {nb_total}")
    else:
        print(f" ALERTE : {nb_q1} + {nb_q2} ≠ {nb_total}")
    
    print("\n" + "="*60)
    print(" EXTRACTION TERMINÉE AVEC SUCCÈS")
    print("="*60)
    
    # 6. Retour du dictionnaire
    return {
        "clients": clients,
        "commandes": commandes,
        "produits": produits
    }
```

```python
# Stocker le résultat dans une variable !
data = extract(
    q1_path=r'C:\Users\Utilisateur\Desktop\projets\ETL_Data_pipeline\data\commandes_Q1.csv',
    q2_path=r'C:\Users\Utilisateur\Desktop\projets\ETL_Data_pipeline\data\commandes_Q2.csv',
    clients_path=r'C:\Users\Utilisateur\Desktop\projets\ETL_Data_pipeline\data\clients.csv',
    produits_path=r'C:\Users\Utilisateur\Desktop\projets\ETL_Data_pipeline\data\produits.csv'
)

# accéder aux DataFrames :
clients_df = data["clients"]
commandes_df = data["commandes"]
produits_df = data["produits"]

# Vérification rapide
print(f"\n Récapitulatif :")
print(f"   - {len(data['clients'])} clients")
print(f"   - {len(data['commandes'])} commandes")
print(f"   - {len(data['produits'])} produits")
```

### Ex T1 ⭐ — Audit initial
Écris une fonction `audit_dataframe(df, nom)` qui affiche :
- Shape
- Nombre de NaN total et par colonne (en %)
- Nombre de doublons
- Types de chaque colonne

```python
def audit_dataframe(df, nom):
    print("="*30)
    print(f" DATAFRAME : {nom.upper()}")
    print("="*30)
    
    # 1. Shape
    print(f"\n Shape : {df.shape[0]} lignes × {df.shape[1]} colonnes")
    
    # 2. NaN - calculé 
    nan_par_colonne = df.isna().sum()
    nan_total = nan_par_colonne.sum()
    nan_pct = (nan_par_colonne / df.shape[0] * 100).round(2)
    
    print(f"\n NaN total : {nan_total}")
    print(f" NaN par colonne (en %) :\n{nan_pct}")
    
    # 3. Doublons
    nb_doublons = df.duplicated().sum()
    print(f"\n Nombre de doublons : {nb_doublons}")
    
    # 4. Types
    print(f"\n Types de données :\n{df.dtypes}")
    print("\n")
    
audit_dataframe(data["commandes"], "commandes")
audit_dataframe(data["clients"], "clients")
audit_dataframe(data["produits"], "produits")
```

### Ex T4 ⭐⭐ — Traiter les NaN
Pour chaque colonne avec des NaN, choisissez et appliquez la stratégie adaptée :
- `montant_total` NaN → recalculer depuis `quantite × prix_unitaire`
- `statut` NaN → valeur par défaut `'inconnu'`
- `date_commande` NaT → supprimer la ligne (donnée critique)

Justifiez chaque choix dans un commentaire.

```python
print("="*40)
print("traitement des  NaN - Commandes")
print("="*40)
print(" Étape 1 : Recalcul de montant_total")
nb_nan_avant = data['commandes']['montant_total'].isna().sum()
data['commandes']['montant_total'] = data['commandes']['montant_total'].fillna(
    data['commandes']['quantite'] * data['commandes']['prix_unitaire']
)
nb_nan_apres = data['commandes']['montant_total'].isna().sum()
print(f"{nb_nan_avant} NaN remplacés ===> {nb_nan_apres} NaN restants")

print(" Étape 2 : Remplacement des statut NaN par 'inconnu'")
nb_nan_avant = data['commandes']['statut'].isna().sum()
data['commandes']['statut'] = data['commandes']['statut'].fillna('inconnu')
nb_nan_apres = data['commandes']['statut'].isna().sum()
print(f"{nb_nan_avant} NaN remplacés → {nb_nan_apres} NaN restants")

print(" Étape 3 : Suppression des lignes avec date_commande ")
nb_lignes_avant = data['commandes'].shape[0]
data['commandes'] = data['commandes'].dropna(subset=['date_commande'])
nb_lignes_apres = data['commandes'].shape[0]
print(f"{nb_lignes_avant - nb_lignes_apres} lignes supprimées")
print(f"Il reste {nb_lignes_apres} lignes")
```

### Ex T5 ⭐⭐ — Détecter et traiter les aberrants
1. Détectez les prix négatifs → supprimez ces lignes
2. Détectez les quantités nulles ou négatives → supprimez
3. Vérifiez la cohérence `montant_total = quantite × prix_unitaire` → corrigez les incohérences
4. Affichez un résumé des corrections effectuées

```python
print("="*40)
print("DÉTECTION ET TRAITEMENT DES VALEURS ABERRANTES")
print("="*40)

# Compter les lignes avant tout traitement
nb_lignes_initial = data["commandes"].shape[0]
print(f"Lignes initiales : {nb_lignes_initial}")

# 1. Détecter et supprimer les prix négatifs
print("\n Étape 1 : Prix négatifs")
nb_prix_negatifs = (data["commandes"]["prix_unitaire"] < 0).sum()
print(f" Détection : {nb_prix_negatifs} prix négatifs trouvés")
data["commandes"] = data["commandes"][data["commandes"]["prix_unitaire"] >= 0]
print(f" Supprimés : {nb_prix_negatifs} lignes")

# 2. Détecter et supprimer les quantités nulles ou négatives
print(" Étape 2 : Quantités nulles ou négatives")
nb_quantite_problematique = (data["commandes"]["quantite"] <= 0).sum()
print(f" Détection : {nb_quantite_problematique} quantités problématiques trouvées")
data["commandes"] = data["commandes"][data["commandes"]["quantite"] > 0]
print(f" Supprimés : {nb_quantite_problematique} lignes")

# 3. Vérifier la cohérence montant_total = quantite × prix_unitaire
print(" Étape 3 : Cohérence montant_total")
# Calculer le montant attendu
data["commandes"]["montant_calcule"] = data["commandes"]["quantite"] * data["commandes"]["prix_unitaire"]

# Détecter les incohérences 
incoherences = abs(data["commandes"]["montant_total"] - data["commandes"]["montant_calcule"]) > 0.01
nb_incoherences = incoherences.sum()
print(f" Détection : {nb_incoherences} incohérences trouvées")

# Corriger les incohérences en recalculant le montant
data["commandes"].loc[incoherences, "montant_total"] = data["commandes"].loc[incoherences, "montant_calcule"]
print(f" Corrigés : {nb_incoherences} montants recalculés")

# Supprimer la colonne temporaire
data["commandes"] = data["commandes"].drop(columns=["montant_calcule"])

# 4. Résumé final
print("\n" + "="*40)
print("RÉSUMÉ DES CORRECTIONS")
print("="*40)
nb_lignes_final = data["commandes"].shape[0]
print(f"   Lignes initiales    : {nb_lignes_initial}")
print(f"   Prix négatifs       : {nb_prix_negatifs} supprimés")
print(f"   Quantités ≤ 0       : {nb_quantite_problematique} supprimés")
print(f"   Incohérences        : {nb_incoherences} corrigées")
print(f"   Lignes finales      : {nb_lignes_final}")
print(f"   Total supprimé      : {nb_lignes_initial - nb_lignes_final} lignes")
print("="*60)
```

### Ex T6 ⭐⭐ — Jointures avec les référentiels
1. Nettoyez `df_clients` (doublons sur `client_id`)
2. Filtrez `df_produits` pour ne garder que les produits actifs
3. Faites un LEFT JOIN commandes ← clients (colonnes : ville, region, segment)
4. Faites un LEFT JOIN commandes ← produits (colonnes : nom, categorie, fournisseur)
5. Vérifiez le nombre de lignes avant et après (doit être identique avec LEFT JOIN)

```python
print("="*40)
print("JOINTURES AVEC LES RÉFÉRENTIELS")
print("="*0)

# 1. Supprimer les doublons dans clients
print("Étape 1 : Nettoyage de clients")
nb_doublons_avant = data['clients'].duplicated().sum()
print(f" Doublons détectés : {nb_doublons_avant}")
data['clients'] = data['clients'].drop_duplicates()
print(f" Doublons supprimés")

# 2. Filtrer les produits actifs
print(" Étape 2 : Filtrage des produits actifs")
nb_produits_avant = data['produits'].shape[0]
data['produits'] = data['produits'][data['produits']['actif'] == True]
nb_produits_apres = data['produits'].shape[0]
print(f"{nb_produits_avant - nb_produits_apres} produits inactifs supprimés")
print(f"{nb_produits_apres} produits actifs restants")

# 3. Sauvegarder le nombre de lignes initial
nb_lignes_initial = data['commandes'].shape[0]
print(f" Lignes dans commandes : {nb_lignes_initial}")

# 4. LEFT JOIN commandes ← clients
print(" Étape 3 : Jointure commandes ← clients")
df_final = pd.merge(
    data['commandes'],
    data['clients'][['client_id', 'ville', 'region', 'segment']],
    on='client_id',
    how='left'
)
print(f" Jointure effectuée : {df_final.shape[0]} lignes")
print(f" Colonnes ajoutées : ville, region, segment")

# 5. LEFT JOIN ← produits
print(" Étape 4 : Jointure ← produits")
df_final = pd.merge(
    df_final,
    data['produits'][['produit_id', 'nom', 'categorie', 'fournisseur']],
    on='produit_id',
    how='left'
)
print(f" Jointure effectuée : {df_final.shape[0]} lignes")
print(f" Colonnes ajoutées : nom, categorie, fournisseur")

# 6. Vérification finale
print("="*40)
print(" VÉRIFICATION FINALE")
print("="*40)
nb_lignes_final = df_final.shape[0]
print(f" Lignes avant jointures : {nb_lignes_initial}")
print(f" Lignes après jointures : {nb_lignes_final}")

if nb_lignes_initial == nb_lignes_final:
    print(f" LEFT JOIN respecté (même nombre de lignes)")
else:
    print(f" Attention : différence de {abs(nb_lignes_initial - nb_lignes_final)} lignes")

print(f" Colonnes finales : {df_final.columns.tolist()}")
print(f" Shape finale : {df_final.shape}")

# Mettre à jour data['commandes'] avec le résultat final
data['commandes'] = df_final
```

### Ex T7 ⭐⭐ — Feature Engineering
Créez les colonnes suivantes :
1. `tranche_montant` : `<100€` / `100-500€` / `500-2000€` / `>2000€` avec `pd.cut()`
2. `saison` : depuis le mois (Hiver/Printemps/Été/Automne)
3. `est_annulee` : booléen True si statut == 'annulée'
4. `ca_perdu` : montant_total si commande annulée, sinon 0
5. `delai_inscription_commande` : nb de jours entre date_inscription du client et date_commande

Affichez `value_counts()` pour les colonnes catégorielles créées.

```python
print("="*40)
print("FEATURE ENGINEERING")
print("="*40)

# ============================================================
# 1. tranche_montant  <100€ / 100-500€ / 500-2000€ / >2000€ avec pd.cut()
# ============================================================
print("Étape 1 : Création de tranche_montant")

# pd.cut() découpe une colonne numérique en intervalles
data['commandes']['tranche_montant'] = pd.cut(
    data['commandes']['montant_total'],
    bins=[-float('inf'), 100, 500, 2000, float('inf')],
    labels=['<100€', '100-500€', '500-2000€', '>2000€']
)
print("Colonne 'tranche_montant' créée")
print(data['commandes']['tranche_montant'].value_counts())

# ============================================================
# 2. saison : mapper les mois vers des saisons
# ============================================================
print("Étape 2 : Création de saison")

# Dictionnaire de mapping mois → saison
mapping_saison = {
    12: 'Hiver', 1: 'Hiver', 2: 'Hiver',
    3: 'Printemps', 4: 'Printemps', 5: 'Printemps',
    6: 'Été', 7: 'Été', 8: 'Été',
    9: 'Automne', 10: 'Automne', 11: 'Automne'
}

# On applique le mapping sur la colonne 'mois' déjà créée dans T3
data['commandes']['saison'] = data['commandes']['mois'].map(mapping_saison)
print("Colonne 'saison' créée")
print(data['commandes']['saison'].value_counts())

# ============================================================
# 3. est_annulee : booléen True si statut == 'annulée'
# ============================================================
print(" Étape 3 : Création de est_annulee")

# Comparaison directe qui retourne True/False
data['commandes']['est_annulee'] = data['commandes']['statut'] == 'annulée'
print("Colonne 'est_annulee' créée")
print(data['commandes']['est_annulee'].value_counts())

# ============================================================
# 4. ca_perdu : montant_total si annulée, sinon 0
# ============================================================
print("Étape 4 : Création de ca_perdu")

import numpy as np

# np.where(condition, valeur_si_vrai, valeur_si_faux)
data['commandes']['ca_perdu'] = np.where(
    data['commandes']['est_annulee'] == True,
    data['commandes']['montant_total'],
    0
)
print("Colonne 'ca_perdu' créée")
print(f" CA perdu total : {data['commandes']['ca_perdu'].sum():.2f} €")

# ============================================================
# 5. delai_inscription_commande : jours entre inscription et commande
# ============================================================
print("Étape 5 : Création de delai_inscription_commande")

# D'abord, s'assurer que date_inscription est en datetime
data['clients']['date_inscription'] = pd.to_datetime(
    data['clients']['date_inscription'], 
    errors='coerce'
)

# Créer un dictionnaire de mapping client_id → date_inscription
dict_inscription = data['clients'].set_index('client_id')['date_inscription'].to_dict()

# Mapper cette date dans commandes
data['commandes']['date_inscription'] = data['commandes']['client_id'].map(dict_inscription)

# Calculer la différence en jours
data['commandes']['delai_inscription_commande'] = (
    data['commandes']['date_commande'] - data['commandes']['date_inscription']
).dt.days

# Supprimer la colonne temporaire
data['commandes'] = data['commandes'].drop(columns=['date_inscription'])

print("Colonne 'delai_inscription_commande' créée")
print(f"Délai moyen : {data['commandes']['delai_inscription_commande'].mean():.1f} jours")
print(f"Délai min : {data['commandes']['delai_inscription_commande'].min()} jours")
print(f"Délai max : {data['commandes']['delai_inscription_commande'].max()} jours")

# ============================================================
# RÉSUMÉ FINAL
# ============================================================
print("="*40)
print("RÉSUMÉ DES NOUVELLES COLONNES CRÉÉES")
print("="*40)
print(f"Shape finale : {data['commandes'].shape}")
print(f"Nouvelles colonnes :")
for col in ['tranche_montant', 'saison', 'est_annulee', 'ca_perdu', 'delai_inscription_commande']:
    print(f"- {col}")
```

### Ex T8 ⭐⭐⭐ — Validation qualité
Écris une fonction `valider_pipeline(df) -> bool` qui :
1. Vérifie qu'il n'y a plus de NaN sur les colonnes critiques
2. Vérifie qu'il n'y a plus de doublons
3. Vérifie que tous les prix sont > 0
4. Vérifie que toutes les quantités sont > 0
5. Vérifie que montant_total == quantite × prix_unitaire (à 1 centime près)
6. Retourne True si tout est OK, False sinon avec les erreurs détaillées

```python
def valider_pipeline(df):
urne True si tout OK, False sinon
    print("="*40)
    print("VALIDATION QUALITÉ DU PIPELINE")
    print("="*40)

    
    # ============================================================
    # 1. Vérifier les NaN sur colonnes critiques
    # ============================================================
    print("Vérification 1 : NaN sur colonnes critiques")
    colonnes_critiques = ['commande_id', 'client_id', 'produit_id', 'montant_total', 
                          'date_commande', 'quantite', 'prix_unitaire']
    
    for col in colonnes_critiques:
        if col in df.columns:
            nb_nan = df[col].isna().sum()
            if nb_nan > 0:
                erreurs.append(f"{nb_nan} NaN dans '{col}'")
                print(f"{nb_nan} NaN dans '{col}'")
            else:
                print(fAucun NaN dans '{col}'")
    
    # ============================================================
    # 2. Vérifier les doublons
    # ============================================================
    print("Vérification 2 : Doublons")
    nb_doublons = df.duplicated().sum()
    if nb_doublons > 0:
        erreurs.append(f"{nb_doublons} doublons détectés")
        print(f"{nb_doublons} doublons détectés")
    else:
        print(f"Aucun doublon")
    
    # ============================================================
    # 3. Vérifier prix > 0
    # ============================================================
    print("Vérification 3 : Prix > 0")
    prix_negatifs = (df['prix_unitaire'] <= 0).sum()
    if prix_negatifs > 0:
        erreurs.append(f"{prix_negatifs} prix ≤ 0 détectés")
        print(f"{prix_negatifs} prix ≤ 0 détectés")
    else:
        print(f"Tous les prix > 0")
    
    # ============================================================
    # 4. Vérifier quantités > 0
    # ============================================================
    print("Vérification 4 : Quantités > 0")
    quantites_negatives = (df['quantite'] <= 0).sum()
    if quantites_negatives > 0:
        erreurs.append(f"{quantites_negatives} quantités ≤ 0 détectées")
        print(f"{quantites_negatives} quantités ≤ 0 détectées")
    else:
        print(f"Toutes les quantités > 0")
    
    # ============================================================
    # 5. Vérifier cohérence montant_total
    # ============================================================
    print("Vérification 5 : Cohérence montant_total")
    montant_calcule = df['quantite'] * df['prix_unitaire']
    incoherences = (abs(df['montant_total'] - montant_calcule) > 0.01).sum()
    
    if incoherences > 0:
        erreurs.append(f"incoherences} incohérences montant_total détectées")
        print(f"{incoherences} incohérences montant_total détectées")
    else:
        print(f"Tous les montants cohérents")
    
    # ============================================================
    # RÉSULTAT FINAL
    # ============================================================
    print("\n" + "="*60)
    if len(erreurs) == 0:
        print("VALIDATION RÉUSSIE ! Toutes les vérifications sont OK")
        print("="*60)
        return True
    else:
        print(f"VALIDATION ÉCHOUÉE ! {len(erreurs)} erreur(s) détectée(s) :")
        for erreur in erreurs:
            print(erreur)
        print("="*60)
        return False
```

### Ex T9 ⭐⭐⭐ — Analyse métier post-transform
Maintenant que les données sont propres, répondez en code :
1. Quel est le CA total du S1 2024 ?
2. Quel trimestre a le CA le plus élevé ?
3. Quelle région génère le plus de CA ?
4. Quel segment client est le plus rentable ?
5. Quelle catégorie de produit a le taux d'annulation le plus élevé ?
6. Quel est le CA perdu sur les annulations ?

```python
print("="*60)
print("T9 : ANALYSE MÉTIER POST-TRANSFORM")
print("="*60)

# On crée un alias pour simplifier l'écriture
df = data['commandes']

# ============================================================
# 1. Quel est le CA total du S1 2024 ?
# ============================================================
# S1 = Semestre 1 = Trimestres 1 et 2
mask_s1 = (df['annee'] == 2024) & (df['trimestre'].isin([1, 2]))
ca_s1 = df.loc[mask_s1, 'montant_total'].sum()
print(f"\n1. CA total du S1 2024 : {ca_s1:,.2f} €")

# ============================================================
# 2. Quel trimestre a le CA le plus élevé ?
# ============================================================
# On groupe par trimestre, on somme le CA, et on prend l'index du max (idxmax)
ca_par_trimestre = df.groupby('trimestre')['montant_total'].sum()
meilleur_trimestre = ca_par_trimestre.idxmax()
print(f"\n2. Trimestre avec le CA le plus élevé : T{meilleur_trimestre} ({ca_par_trimestre.max():,.2f} €)")

# ============================================================
# 3. Quelle région génère le plus de CA ?
# ============================================================
ca_par_region = df.groupby('region')['montant_total'].sum()
meilleure_region = ca_par_region.idxmax()
print(f"\n3. Région générant le plus de CA : {meilleure_region} ({ca_par_region.max():,.2f} €)")

# ============================================================
# 4. Quel segment client est le plus rentable ?
# ============================================================
ca_par_segment = df.groupby('segment')['montant_total'].sum()
meilleur_segment = ca_par_segment.idxmax()
print(f"\n4. Segment client le plus rentable : {meilleur_segment} ({ca_par_segment.max():,.2f} €)")

# ============================================================
# 5. Quelle catégorie de produit a le taux d'annulation le plus élevé ?
# ============================================================
# Astuce : est_annulee est un booléen (True=1, False=0). 
# La moyenne (mean) d'une colonne booléenne donne directement le taux (pourcentage) !
taux_annulation_par_cat = df.groupby('categorie')['est_annulee'].mean() * 100
pire_categorie = taux_annulation_par_cat.idxmax()
print(f"\n 5. Catégorie avec le taux d'annulation le plus élevé : {pire_categorie} ({taux_annulation_par_cat.max():.2f}%)")

# ============================================================
# 6. Quel est le CA perdu sur les annulations ?
# ============================================================
# On utilise la colonne 'ca_perdu' créée dans l'exercice T7
ca_perdu_total = df['ca_perdu'].sum()
print(f"\n6. CA perdu sur les annulations : {ca_perdu_total:,.2f} €")

print("\n" + "="*60)
print("ANALYSE MÉTIER TERMINÉE")
print("="*60)
```
